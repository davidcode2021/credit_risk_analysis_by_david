```python
import pandas as pd 
import numpy as np 
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.ensemble import RandomForestRegressor
import xgboost as xgb
import os 
```

```python
os.getcwd()
```

```python
df_previous = pd.read_csv("data/raw/previous_app.csv")
```

```python
# converting all numerical data into 32-bit t0 reduce memory usage 

# Integer columns
int_columns = df_previous.select_dtypes(include='int').columns.tolist()
# Float columns
float_columns = df_previous.select_dtypes(include='float').columns.tolist()

# Convert integer columns to int32
df_previous[int_columns] = df_previous[int_columns].astype("int32")

# Convert float columns to float32
df_previous[float_columns] = df_previous[float_columns].astype("float32")
```

```python
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
```

```python
df_previous.head()
```

```python
# df_previous.duplicated().any()

missing_pct=df_previous.isna().mean() * 100
missing_pct=missing_pct[missing_pct > 0].sort_values(ascending=False)

missing_pct
```

```python
cols_to_drop = missing_pct.index[0:2].to_list() + ["SK_ID_PREV"]
cols_for_advance_impute = missing_pct.index[2:14].to_list()
cols_simple_impute = missing_pct.index[-2:].to_list()

df_previous = df_previous.drop(cols_to_drop, axis=1)
df_previous[cols_simple_impute[0]] = df_previous[cols_simple_impute[0]].fillna(df_previous[cols_simple_impute[0]].mode().iloc[0])
df_previous[cols_simple_impute[1]] = df_previous[cols_simple_impute[1]].fillna(df_previous[cols_simple_impute[1]].median())

for col in cols_for_advance_impute:
    cat_cols = []
    num_cols = {}
    if df_previous[col].dtype.name == 'object':
        
        value_counts=df_previous[col].value_counts()
        ratios = value_counts / value_counts.sum() #calculating ratio for each category
        
        nan_counts= df_previous[col].isna().sum()  #counting nan value for filling
        
        fill_values = np.random.choice(
        ratios.index,           # The category names
        size=nan_counts,         # How many values to generate
        p=ratios.values         # The probabilities/ratios
        )
        df_previous.loc[df_previous[col].isna(), col] = fill_values  #Fill the NaN positions with these random values
        
    else:
        
        # MICE method of filling
        imputer = IterativeImputer(
        estimator=RandomForestRegressor(n_estimators=5, max_depth=4, random_state=42, n_jobs=-1),           
        max_iter=5, random_state=42)
        df_previous[[col]] = imputer.fit_transform(df_previous[[col]]) 
    
```

```python
"""
checking for  columns that maybe corrrelated
"""
df_previous_numeric = df_previous.select_dtypes(include=["number"])

corr_matrix= df_previous_numeric.corr()
# Unstack the matrix into pairs
corr_pairs = corr_matrix.unstack()

# Remove self-correlations
corr_pairs = corr_pairs[corr_pairs != 1]

# Keep only strong correlations
strong_corr = corr_pairs[abs(corr_pairs) > 0.8]

# Sort by strength
df_strong_corr = strong_corr.sort_values(key=abs, ascending=False).reset_index()
corr_cols=df_strong_corr[["level_0"]].iloc[::2]

corr_cols=corr_cols.values.flatten().tolist()

# dropping all the correlated columns 
df_previous = df_previous.drop(corr_cols, axis=1)
```

```python
# saving the clean data frame  into precessed data folder
# df_previous.to_csv("~/Desktop/credit_analysis/data/processed/df_previous_clean")
```

```python
TARGET = pd.read_csv("data/raw/current_app.csv")[["TARGET", "SK_ID_CURR"]]   #target of the current data 
merged_previous = df_previous.merge( TARGET, on="SK_ID_CURR", how= "left")
merged_previous = merged_previous.dropna() #droping all the nan values 
```

```python
merged_previous.head()
```

```python
import pandas as pd
from scipy.stats import ttest_ind, chi2_contingency

numeric_list = []
cats_list = []

for col in merged_previous.columns:

    if col == "TARGET":
        continue

    # Numerical features
    if pd.api.types.is_numeric_dtype(merged_previous[col]):

        temp = merged_previous[[col, "TARGET"]].dropna()

        group0 = temp[temp["TARGET"] == 0][col]
        group1 = temp[temp["TARGET"] == 1][col]

        if len(group0) > 1 and len(group1) > 1:
            _, p_value = ttest_ind(
                group0,
                group1,
                equal_var=False  # Welch's t-test
            )

            numeric_list.append({
                "column": col,
                "p_value": p_value
            })

    # Categorical features
    else:

        temp = merged_previous[[col, "TARGET"]].dropna()

        contingency = pd.crosstab(
            temp[col],
            temp["TARGET"]
        )

        _, p_value, _, _ = chi2_contingency(contingency)

        cats_list.append({
            "column": col,
            "p_value": p_value
        })

numeric_df = (
    pd.DataFrame(numeric_list)
    .sort_values("p_value")
)

cats_df = (
    pd.DataFrame(cats_list)
    .sort_values("p_value")
)


cats_df.head()
# numeric_df.head()
```

The statistical test conducted in the cell above is just to sieve out the important features  with good predictitve power this is so in other to reduce the number of features. Only features with high predictive power will be aggregated and merged to the curren data. However from the result, all the p_values are zeros. Showing all are importants, that means I may not be able to know the most important features. The better approached is to use tree base model to indentify the most important features.

```python

cat_cols = merged_previous.select_dtypes(include=["object"]).columns
merged_previous[cat_cols] = merged_previous[cat_cols].astype("category")

X_train = merged_previous.drop(["SK_ID_CURR","TARGET"], axis=1)
y_train = merged_previous["TARGET"]

model = xgb.XGBClassifier(n_estimators=100, 
        random_state=42,enable_categorical=True)

model.fit(X_train, y_train)

importance_df = pd.DataFrame({
    'Feature': X_train.columns.tolist(),
    'Importance': model.feature_importances_
}).sort_values(by='Importance', ascending=False)
```

```python
important_features = importance_df.reset_index(drop=True)

# picking only the five most important features
important_five =important_features["Feature"][:5]
important_five
```
