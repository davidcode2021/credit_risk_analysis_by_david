```python
final_df = df_current.merge(df_previous, on ="SK_ID_CURR", how = "inner")
final_df["TARGET"].value_counts()
```

i will then build an XGboost model that will predict how risky each customer is,

```python
xgb_params = {
    "objective": "binary:logistic",
    "eval_metric": "logloss",
    "learning_rate": 0.1,      
    "max_depth": 6,               
    "gamma": 0.0,                
    "subsample": 0.8,             
    "colsample_bytree": 0.8,      
    "reg_alpha": 0.1,           
    "reg_lambda": 1.0,            
    "scale_pos_weight": 1.0,      #since the data is not balanced 
    "random_state": 42,
    "n_jobs": -1                  # Use all available CPU cores
}

X = final_df.drop(["TARGET","SK_ID_CURR"], axis=1)
cat_cols = X.select_dtypes(include=['object']).columns
X[cat_cols] = X[cat_cols].astype('category')
y = final_df["TARGET"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=None)

model = XGBClassifier(n_estimators= 700,enable_categorical=True,**xgb_params)
model.fit(X_train, y_train, eval_set=[(X_test, y_test)], verbose=False)

y_probs = model.predict_proba(X_test)[:, 1]

# Convert probabilities to 0 or 1 using a 0.5 threshold
y_preds = (y_probs >= 0.5).astype(int)

# --- METRICS ---
# Standard Accuracy (% of correct predictions)
acc = accuracy_score(y_test, y_preds)
print(f"Accuracy: {acc:.4f}")
```

```python
final_df["TARGET_probs"] = model.predict_proba(X)[:, 1]
bins = [0.0, 0.3, 0.7, 1.0]
labels = ['low', 'medium', 'high']
# Create the new column
final_df['risk_level'] = pd.cut(final_df['TARGET_probs'], bins=bins, labels=labels, include_lowest=True)
```

### ANALYSIS

## Applicant profile information base on the the risk level:

```python
from IPython.display import display
profil_cols = ["CODE_GENDER","NAME_FAMILY_STATUS","NAME_EDUCATION_TYPE"]

# checking whether there is an association between risk_level and the columns in profil_cols 

def cramers_v(contingency_table):
    chi2 = chi2_contingency(contingency_table)[0]
    n = contingency_table.sum().sum()
    min_dim = min(contingency_table.shape) - 1
    if min_dim == 0:
        return 0
    return np.sqrt(chi2 / (n * min_dim))

dic_list= []
for cols in profil_cols:
    value_table = pd.crosstab(final_df["risk_level"], final_df[cols])
    _, p_value, _, _ = chi2_contingency(value_table)
    c_v = cramers_v(value_table)
    dic_list.append({
    "profil_cols": cols,
    "p_values": p_value,
     "cramers_v": round(c_v, 4)   
    })
result= pd.DataFrame(dic_list).sort_values(by="p_values", ascending= False)

print("p_value for each profile columns : " )
display(result)


info_table1 = pd.get_dummies(final_df[profil_cols]).groupby(final_df["risk_level"], observed= False).sum()
info_table1 = info_table1.reset_index()
info_table1.columns = ["risk_level","females","male","unknown", "civil_marriage", "Married","saparated","single","widows",
                       "academic_degree", "higher_education", "incomplete_higher","lower_secondary", "secondary"]

info_table1 = info_table1.drop("unknown", axis=1)

gender_cols = ["females", "male"]

gender_pct = info_table1[["risk_level"] + gender_cols].copy()

gender_pct[gender_cols] = (
    gender_pct[gender_cols]
    .div(gender_pct[gender_cols].sum(axis=1), axis=0)
    * 100
).round(2)

family_cols = ["civil_marriage","Married","saparated","single","widows"]
family_pct = info_table1[["risk_level"] + family_cols].copy()

family_pct[family_cols] = (
    family_pct[family_cols]
    .div(family_pct[family_cols].sum(axis=1), axis=0)
    * 100
).round(2)

edu_cols = ["academic_degree", "higher_education","incomplete_higher","lower_secondary","secondary"]

edu_pct = info_table1[["risk_level"] + edu_cols].copy()

edu_pct[edu_cols] = (
    edu_pct[edu_cols]
    .div(edu_pct[edu_cols].sum(axis=1), axis=0)
    * 100
).round(2)
```

## Loan Characteristics base on the risk level 

```python
loan_xteristics_cols = ["NAME_CONTRACT_TYPE", "CNT_PAYMENT_median","DLDV_median"]

# Helper for Eta-Squared (Effect size for ANOVA / continuous variables)
def eta_squared(groups):
    all_data = np.concatenate(groups)
    grand_mean = np.mean(all_data)
    sst = np.sum((all_data - grand_mean) ** 2)
    ssb = np.sum(
        [len(g) * (np.mean(g) - grand_mean) ** 2 for g in groups if len(g) > 0]
    )
    return ssb / sst if sst != 0 else 0


# Statistical Tests & Association Table ---
dic_list = []

for cols in loan_xteristics_cols:
    if final_df[cols].dtype == "object" or final_df[cols].nunique() <= 10:
        # Categorical feature -> Chi-Square Test & Cramer's V
        value_table = pd.crosstab(final_df["risk_level"], final_df[cols])
        _, p_value, _, _ = chi2_contingency(value_table)
        effect_size = round(cramers_v(value_table), 4)
        stat_metric = "Cramer's V"
    else:
        # Numerical/Continuous feature -> One-Way ANOVA & Eta-Squared
        groups = [
            group[cols].dropna().values
            for _, group in final_df.groupby("risk_level", observed=False)
        ]
        _, p_value = f_oneway(*groups)
        effect_size = round(eta_squared(groups), 4)
        stat_metric = "Eta-Squared"

    dic_list.append(
        {
            "profil_cols": cols,
            "p_values": p_value,
            "effect_size": effect_size,
            "effect_type": stat_metric,
        }
    )

result = pd.DataFrame(dic_list).sort_values(by="p_values", ascending=False)
print("Statistical summary for loan characteristics columns:")
display(result)



contract_type_pct = (
    pd.crosstab(
        final_df["risk_level"], final_df["NAME_CONTRACT_TYPE"], normalize="index"
    )
    * 100
).round(2)
contract_type_pct = contract_type_pct.reset_index()

#  Numerical: CNT_PAYMENT & DLDV_median (Mean/Median by risk_level)
loan_num_stats = (
    final_df.groupby("risk_level", observed= False)[["CNT_PAYMENT_median", "DLDV_median"]]
    .agg(["mean"])
    .round(2)
)
loan_num_stats.columns = [
    f"{col}_{stat}" for col, stat in loan_num_stats.columns
]
loan_num_stats = loan_num_stats.reset_index()
    
```
